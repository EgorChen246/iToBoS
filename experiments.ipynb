# iToBoS: сравнение моделей
Результаты теста: 30 фото, 116 объектов. Метрики вычислены при заранее выбранных порогах; повторное обучение здесь не запускается.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
root = Path.cwd()
df = pd.read_json(root / "results" / "test_metrics.json")
display(df[["model", "threshold", "Precision", "Recall", "F1", "MeanIoU_TP", "CountMAE", "CountExactAccuracy", "warm_mean_seconds"]])
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
df.set_index("model")[["Precision", "Recall", "F1"]].plot.bar(ax=axes[0], ylim=(0, 1), rot=15)
axes[0].set_title("Качество на тестовой выборке")
df.set_index("model")["warm_mean_seconds"].plot.bar(ax=axes[1], rot=15)
axes[1].set_ylabel("Секунды / фото, CPU")
axes[1].set_title("После первой обработки; без сети")
plt.tight_layout()
plt.show()
```
